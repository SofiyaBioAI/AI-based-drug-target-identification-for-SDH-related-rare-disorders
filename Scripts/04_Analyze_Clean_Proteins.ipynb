```python
import os
import pandas as pd
from Bio.PDB import PDBParser
```

```python
project_folder = r"C:\Users\sofiya chavarekar\OneDrive\Assignment- Module 1\PROJECT(AI-based drug target identification for SDH-related rare disorders)"

clean_folder = os.path.join(project_folder, "Data", "Clean_Proteins")
result_folder = os.path.join(project_folder, "Data", "Results")

os.makedirs(result_folder, exist_ok=True)
```

```python
parser = PDBParser(QUIET=True)

protein_data = []

files = ["6C12", "2H89", "1NEN", "1Y8P"]

for pdb in files:

    structure = parser.get_structure(
        pdb,
        os.path.join(clean_folder, f"{pdb}_clean.pdb")
    )

    atoms = list(structure.get_atoms())
    residues = list(structure.get_residues())
    chains = list(structure.get_chains())

    protein_data.append({
        "Protein": pdb,
        "Chains": len(chains),
        "Residues": len(residues),
        "Atoms": len(atoms)
    })

df = pd.DataFrame(protein_data)

df
```

```python
output_file = os.path.join(result_folder, "Clean_Protein_Analysis.xlsx")

df.to_excel(output_file, index=False)

print("Analysis saved successfully!")
print(output_file)
```
