```python
import os
import csv
import subprocess

# =====================================================
# PATHS
# =====================================================

VINA = r"C:\Users\sofiya chavarekar\Downloads\vina_1.2.7_win.exe"

RECEPTOR = r"C:\Users\sofiya chavarekar\OneDrive\Assignment- Module 1\PROJECT(AI-based drug target identification for SDH-related rare disorders)\Data\PDBQT_Proteins\6C12_receptor.pdbqt"

LIGAND_FOLDER = r"C:\Users\sofiya chavarekar\OneDrive\Assignment- Module 1\PROJECT(AI-based drug target identification for SDH-related rare disorders)\Data\Ligands\PDBQT"

OUTPUT_FOLDER = r"C:\Users\sofiya chavarekar\OneDrive\Assignment- Module 1\PROJECT(AI-based drug target identification for SDH-related rare disorders)\Docking\6C12"

RESULT_FILE = r"C:\Users\sofiya chavarekar\OneDrive\Assignment- Module 1\PROJECT(AI-based drug target identification for SDH-related rare disorders)\Results\6C12_results.csv"

# =====================================================
# GRID BOX
# =====================================================

CENTER_X = -15.669
CENTER_Y = 2.562
CENTER_Z = 25.740

SIZE_X = 20
SIZE_Y = 20
SIZE_Z = 20

CPU = 4

# =====================================================
# CHECK PATHS
# =====================================================

print("=" * 60)
print("CHECKING PATHS")
print("=" * 60)

print("Vina      :", os.path.exists(VINA))
print("Receptor  :", os.path.exists(RECEPTOR))
print("Ligands   :", os.path.exists(LIGAND_FOLDER))
print("Results   :", os.path.exists(os.path.dirname(RESULT_FILE)))

if not os.path.exists(VINA):
    raise Exception("Vina executable not found.")

if not os.path.exists(RECEPTOR):
    raise Exception("Receptor not found.")

if not os.path.exists(LIGAND_FOLDER):
    raise Exception("Ligand folder not found.")

os.makedirs(OUTPUT_FOLDER, exist_ok=True)
os.makedirs(os.path.dirname(RESULT_FILE), exist_ok=True)

# =====================================================
# GET LIGANDS
# =====================================================

ligands = sorted([
    f for f in os.listdir(LIGAND_FOLDER)
    if f.endswith(".pdbqt")
])

print(f"\nFound {len(ligands)} ligands.\n")

# =====================================================
# CREATE CSV
# =====================================================

with open(RESULT_FILE, "w", newline="") as csvfile:

    writer = csv.writer(csvfile)
    writer.writerow(["Ligand", "Affinity (kcal/mol)"])

    for i, ligand in enumerate(ligands, start=1):

        ligand_path = os.path.join(LIGAND_FOLDER, ligand)

        output_file = os.path.join(
            OUTPUT_FOLDER,
            ligand.replace(".pdbqt", "_out.pdbqt")
        )

        cmd = [
            VINA,
            "--receptor", RECEPTOR,
            "--ligand", ligand_path,
            "--center_x", str(CENTER_X),
            "--center_y", str(CENTER_Y),
            "--center_z", str(CENTER_Z),
            "--size_x", str(SIZE_X),
            "--size_y", str(SIZE_Y),
            "--size_z", str(SIZE_Z),
            "--cpu", str(CPU),
            "--out", output_file
        ]

        print(f"[{i}/{len(ligands)}] Docking {ligand}...")

        try:

            result = subprocess.run(
                cmd,
                capture_output=True,
                text=True
            )

            affinity = "NA"

            if result.returncode == 0:

                for line in result.stdout.splitlines():

                    cols = line.split()

                    if len(cols) >= 2 and cols[0].isdigit():
                        affinity = cols[1]
                        break

            else:

                print("Docking failed:")
                print(result.stderr)

            writer.writerow([ligand, affinity])

            print("Affinity =", affinity)

        except Exception as e:

            print("Error:", e)
            writer.writerow([ligand, "ERROR"])

print("\n" + "=" * 60)
print("DOCKING FINISHED")
print("=" * 60)
print("Results saved to:")
print(RESULT_FILE)
print("Docked structures saved in:")
print(OUTPUT_FOLDER)
```

```python
import os
import csv
import subprocess

# -------------------------------
# PATHS
# -------------------------------

VINA = r"C:\Users\sofiya chavarekar\Downloads\vina_1.2.7_win.exe"

PROJECT = r"C:\Users\sofiya chavarekar\OneDrive\Assignment- Module 1\PROJECT(AI-based drug target identification for SDH-related rare disorders)"

RECEPTOR = os.path.join(
    PROJECT,
    "Data",
    "PDBQT_Proteins",
    "6C12_receptor.pdbqt"
)

LIGAND_FOLDER = os.path.join(
    PROJECT,
    "Data",
    "Ligands",
    "PDBQT"
)

OUTPUT_FOLDER = os.path.join(
    PROJECT,
    "Docking",
    "6C12"
)

RESULT_FILE = os.path.join(
    PROJECT,
    "Results",
    "6C12_results.csv"
)

os.makedirs(OUTPUT_FOLDER, exist_ok=True)

# -------------------------------
# GRID PARAMETERS
# -------------------------------

CENTER_X = -15.669
CENTER_Y = 2.562
CENTER_Z = 25.740

SIZE_X = 20
SIZE_Y = 20
SIZE_Z = 20

# -------------------------------
# FIND LIGANDS
# -------------------------------

ligands = sorted(
    f for f in os.listdir(LIGAND_FOLDER)
    if f.endswith(".pdbqt")
)

print(f"Found {len(ligands)} ligands.")

results = []

# -------------------------------
# DOCKING
# -------------------------------

for i, ligand in enumerate(ligands, start=1):

    ligand_path = os.path.join(LIGAND_FOLDER, ligand)

    output_path = os.path.join(
        OUTPUT_FOLDER,
        ligand.replace(".pdbqt", "_out.pdbqt")
    )

    cmd = [
        VINA,
        "--receptor", RECEPTOR,
        "--ligand", ligand_path,

        "--center_x", str(CENTER_X),
        "--center_y", str(CENTER_Y),
        "--center_z", str(CENTER_Z),

        "--size_x", str(SIZE_X),
        "--size_y", str(SIZE_Y),
        "--size_z", str(SIZE_Z),

        "--cpu", "4",
        "--exhaustiveness", "8",

        "--out", output_path
    ]

    result = subprocess.run(
        cmd,
        capture_output=True,
        text=True
    )

    affinity = "NA"

    for line in result.stdout.splitlines():

        parts = line.split()

        if len(parts) >= 2 and parts[0] == "1":
            affinity = parts[1]
            break

    results.append([ligand, affinity])

    print(f"{i}/{len(ligands)}  {ligand}  {affinity}")

# -------------------------------
# SAVE CSV
# -------------------------------

with open(RESULT_FILE, "w", newline="") as f:

    writer = csv.writer(f)

    writer.writerow(["Ligand", "Affinity (kcal/mol)"])

    writer.writerows(results)

print("\nFinished!")
print("Results saved to:")
print(RESULT_FILE)
```

```python
from rdkit import Chem
import pandas as pd
import os

sdf_file = r"C:\Users\sofiya chavarekar\OneDrive\Assignment- Module 1\PROJECT(AI-based drug target identification for SDH-related rare disorders)\Data\Ligands\Raw\drug_like_molecules.sdf"

supplier = Chem.SDMolSupplier(sdf_file)

rows = []

for i, mol in enumerate(supplier):
    if mol is None:
        continue

    ligand_id = f"Ligand_{i+1:04d}"
    name = mol.GetProp("_Name") if mol.HasProp("_Name") else ""
    smiles = Chem.MolToSmiles(mol)

    rows.append([ligand_id, name, smiles])

df = pd.DataFrame(rows, columns=["Ligand", "Chemical_Name", "SMILES"])

output = r"C:\Users\sofiya chavarekar\OneDrive\Assignment- Module 1\PROJECT(AI-based drug target identification for SDH-related rare disorders)\Analysis\All_Ligands_SMILES.csv"

os.makedirs(os.path.dirname(output), exist_ok=True)
df.to_csv(output, index=False)

print(df.head())
print(f"\nSaved {len(df)} ligands to:")
print(output)
```
