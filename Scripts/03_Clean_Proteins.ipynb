```python
import os
from Bio.PDB import PDBParser, PDBIO, Select
```

```python
# Project folder
project_folder = r"C:\Users\sofiya chavarekar\OneDrive\Assignment- Module 1\PROJECT(AI-based drug target identification for SDH-related rare disorders)"
protein_folder = os.path.join(project_folder, "Data", "Proteins")

clean_folder = os.path.join(project_folder, "Data", "Clean_Proteins")

os.makedirs(clean_folder, exist_ok=True)
```

```python
class ProteinSelect(Select):
    def accept_residue(self, residue):
        # Remove water molecules
        if residue.get_resname() == "HOH":
            return 0

        # Keep only standard amino acids
        return residue.id[0] == " "
```

```python
parser = PDBParser(QUIET=True)
io = PDBIO()

files = ["6C12", "2H89", "1NEN", "1Y8P"]

for pdb in files:

    structure = parser.get_structure(
        pdb,
        os.path.join(protein_folder, f"{pdb}.pdb")
    )

    io.set_structure(structure)

    output_file = os.path.join(
        clean_folder,
        f"{pdb}_clean.pdb"
    )

    io.save(output_file, ProteinSelect())

    print(f"{pdb} cleaned successfully")

print("\nAll proteins cleaned!")
```
