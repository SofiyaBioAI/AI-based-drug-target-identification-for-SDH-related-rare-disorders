```python
import os
import requests

# Project folders
project_folder = r"C:\Users\sofiya chavarekar\OneDrive\Assignment- Module 1\PROJECT(AI-based drug target identification for SDH-related rare disorders"

protein_folder = os.path.join(project_folder, "Data", "Proteins")
os.makedirs(protein_folder, exist_ok=True)

# Protein PDB IDs
proteins = {
    "SDHA": "6C12",
    "SDHB": "2H89",
    "SDHC": "1NEN",
    "SDHD": "1Y8P"
}

# Download proteins
for protein, pdb_id in proteins.items():

    url = f"https://files.rcsb.org/download/{pdb_id}.pdb"

    response = requests.get(url)

    if response.status_code == 200:

        filename = os.path.join(protein_folder, f"{pdb_id}.pdb")

        with open(filename, "w") as file:
            file.write(response.text)

        print(f"{protein} downloaded successfully")

    else:
        print(f"Failed to download {protein}")

print("\nAll downloads completed.")
```
