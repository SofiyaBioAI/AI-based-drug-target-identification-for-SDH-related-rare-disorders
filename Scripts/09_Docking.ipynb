```python
import os
import subprocess

# ==========================================
# AutoDock Vina executable
# ==========================================
vina = r"C:\Users\sofiya chavarekar\Downloads\vina_1.2.7_win.exe"

# ==========================================
# Receptor (YOUR VERIFIED PATH)
# ==========================================
receptor = r"C:\Users\sofiya chavarekar\OneDrive\Assignment- Module 1\PROJECT(AI-based drug target identification for SDH-related rare disorders)\Data\Proteins\6C12_clean.pdbqt"

# ==========================================
# Ligand
# ==========================================
ligand = r"C:\Users\sofiya chavarekar\OneDrive\Assignment- Module 1\PROJECT(AI-based drug target identification for SDH-related rare disorders)\Data\Ligands\PDBQT\Ligand_0001.pdbqt"

# ==========================================
# Output folder
# ==========================================
output_folder = r"C:\Users\sofiya chavarekar\OneDrive\Assignment- Module 1\PROJECT(AI-based drug target identification for SDH-related rare disorders)\Docking\SDHA"

os.makedirs(output_folder, exist_ok=True)

output_file = os.path.join(output_folder, "Ligand_0001_out.pdbqt")

# ==========================================
# Check files
# ==========================================
print("Checking files...\n")

print("Vina exists      :", os.path.exists(vina))
print("Receptor exists  :", os.path.exists(receptor))
print("Ligand exists    :", os.path.exists(ligand))

if not os.path.exists(vina):
    raise FileNotFoundError("Vina executable not found!")

if not os.path.exists(receptor):
    raise FileNotFoundError("Receptor file not found!")

if not os.path.exists(ligand):
    raise FileNotFoundError("Ligand file not found!")

# ==========================================
# Run Vina
# ==========================================
command = [
    vina,
    "--receptor", receptor,
    "--ligand", ligand,
    "--center_x", "-15.669",
    "--center_y", "2.562",
    "--center_z", "25.740",
    "--size_x", "20",
    "--size_y", "20",
    "--size_z", "20",
    "--cpu", "4",
    "--out", output_file
]

print("\nRunning AutoDock Vina...\n")

result = subprocess.run(command, capture_output=True, text=True)

print("=" * 60)
print("RETURN CODE:", result.returncode)
print("=" * 60)

print("\nSTDOUT:\n")
print(result.stdout)

print("\nSTDERR:\n")
print(result.stderr)

print("\nOutput file exists:", os.path.exists(output_file))

if os.path.exists(output_file):
    print("\n✅ Docking completed successfully!")
    print(output_file)
else:
    print("\n❌ Docking failed.")
```
