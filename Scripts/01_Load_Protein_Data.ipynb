```python
from Bio import SeqIO
from Bio.SeqUtils.ProtParam import ProteinAnalysis
import pandas as pd
import os

protein_folder = r"../Data/Proteins"

results = []

for file in os.listdir(protein_folder):

    if file.endswith(".fasta"):

        record = SeqIO.read(os.path.join(protein_folder, file), "fasta")

        sequence = str(record.seq)

        analysis = ProteinAnalysis(sequence)

        results.append({

            "Protein": file.replace(".fasta",""),
            "Length": len(sequence),
            "Molecular Weight": round(analysis.molecular_weight(),2),
            "Isoelectric Point": round(analysis.isoelectric_point(),2),
            "Aromaticity": round(analysis.aromaticity(),3),
            "Instability Index": round(analysis.instability_index(),2),
            "GRAVY": round(analysis.gravy(),3)

        })

df = pd.DataFrame(results)

df
```

```python
import os

output_folder = r"../Data/Results"

print("Output Folder Exists:", os.path.exists(output_folder))
print("Saving files to:", os.path.abspath(output_folder))
```

```python
df.to_excel(
    os.path.join(output_folder, "Protein_Properties.xlsx"),
    index=False
)

print("Protein properties saved successfully!")
```
