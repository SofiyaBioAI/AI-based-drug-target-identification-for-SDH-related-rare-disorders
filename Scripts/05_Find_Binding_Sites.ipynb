```python
import os
import pandas as pd
from Bio.PDB import PDBParser
```

```python
project_folder = r"C:\Users\sofiya chavarekar\OneDrive\Assignment- Module 1\PROJECT(AI-based drug target identification for SDH-related rare disorders)"

clean_folder = os.path.join(project_folder, "Data", "Clean_Proteins")
result_folder = os.path.join(project_folder, "Data", "Results")
```

```python
parser = PDBParser(QUIET=True)

proteins = [
    "6C12",
    "2H89",
    "1NEN",
    "1Y8P"
]

binding_sites = []

for protein in proteins:

    structure = parser.get_structure(
        protein,
        os.path.join(clean_folder, protein + "_clean.pdb")
    )

    atoms = list(structure.get_atoms())

    x = sum(atom.coord[0] for atom in atoms) / len(atoms)
    y = sum(atom.coord[1] for atom in atoms) / len(atoms)
    z = sum(atom.coord[2] for atom in atoms) / len(atoms)

    binding_sites.append({
        "Protein": protein,
        "Center_X": round(x,3),
        "Center_Y": round(y,3),
        "Center_Z": round(z,3)
    })

df = pd.DataFrame(binding_sites)

df
```

```python
output = os.path.join(
    result_folder,
    "Binding_Site_Centers.xlsx"
)

df.to_excel(output,index=False)

print("Binding site coordinates saved successfully!")
print(output)
```
